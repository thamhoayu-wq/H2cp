The task is to write a function that takes a sequence of characters that represents a quantity of data and unit, and translates this quantity to a different unit.

The basic unit of data is the byte (B):

- A kilobyte (KB) is 10<sup>3</sup> bytes
- A megabyte (MB) is 10<sup>6</sup> bytes
- A gigabyte (GB) is 10<sup>9</sup> bytes
- A terabyte (TB) is 10<sup>12</sup> bytes.

For example, 8KB has 8000 bytes.

**Task 3.1**

Write a function called `task3_1(quantity_of_data)` that:

- takes a string, `quantity_of_data`
- tests that the given string is a sequence of digits followed by one of the four approved units shown above (KB, MB, GB, TB).
- returns and displays either:
  - the actual number of bytes represented by the input string

  or
  - the error message, `"invalid data"`. **[5]**

Test the function fully with suitable test data, including all four approved units.

For example,

`task3_1("8KB")`

should return and display `8000` **[3]**


````python
# Task 3.1
def task3_1(quantity_of_data):
  value = quantity_of_data[:-2]
  unit = quantity_of_data[-2:]
  multiplier = 0

  if unit == "KB":
    multiplier = 10**3
  elif unit == "MB":
    multiplier = 10**6
  elif unit == "GB":
    multiplier = 10**9
  elif unit == "TB":
    multiplier = 10**12
  else:
    return "Invalid data"

  output = int(value) * multiplier
  return output

# main
print(task3_1("8KB"))
print(task3_1("8AS"))
print(task3_1("8S"))
````

**Task 3.2**

Companion units are also defined in terms of powers of 2. These have similar abbreviations, as shown:

- A kibibyte (KiB) is 2<sup>10</sup> bytes
- A mebibyte (MiB) is 2<sup>20</sup> bytes
- A gibibyte (GiB) is 2<sup>30</sup> bytes
- A tebibyte (TiB) is 2<sup>40</sup> bytes.

Write a second function `task3_2(quantity_of_data)` that:

- takes a string, `quantity_of_data`
- tests that the given string is a sequence of digits followed by one of the eight approved units (KB, KiB, MB, MiB, GB, GiB, TB, TiB)
- returns and displays either:
  - the number of bytes represented by the input string

  or
  - the error message, `"invalid data"` **[5]**

Test the function fully with suitable test data, including all eight approved units.

For example,

`task3_2("2MiB")`

should return and display `2097152` **[3]**


````python
# Task 3.2
def task3_2(quantity_of_data):
  value = quantity_of_data[:-2]
  unit = quantity_of_data[-2:]
  multiplier = 0

  if unit[0] == "i":
    unit = quantity_of_data[-3:]
    value = quantity_of_data[:-3]

  if unit == "KB":
    multiplier = 10**3
  elif unit == "MB":
    multiplier = 10**6
  elif unit == "GB":
    multiplier = 10**9
  elif unit == "TB":
    multiplier = 10**12
  elif unit == "TiB":
    multiplier = 2**10
  elif unit == "MiB":
    multiplier = 2**20
  elif unit == "GiB":
    multiplier = 2**30
  elif unit == "TiB":
    multiplier = 2**40
  else:
    return "Invalid data"

  output = int(value) * multiplier
  return output

# main
print(task3_2("2MiB"))
print(task3_2("2Mi"))
print(task3_2("2MB"))
````

**Task 3.3**

Write a third function, `task3_3(quantity_of_data, target_unit)` that:

- takes two strings, `quantity_of_data` and `target_unit`
- tests that `target_unit` is one of the eight approved units from task 3.2
- uses your function `task3_2` to generate the actual number of bytes represented by `quantity_of_data`
- converts the generated number of bytes into `target_unit`
- returns and displays either:
  - the `quantity_of_data` in terms of the `target_unit`

  or
  - the error message, `"invalid data"` **[4]**

Test the function with **three** suitable sets of values.

For example,

`task3_3("512MiB", "GiB")`

should return and display `0.5` **[3]**


````python
# Task 3.3
def task3_3(quantity_of_data, target_unit):
  unit = target_unit
  convert = 0

  if unit == "KB":
    convert = 10**3
  elif unit == "MB":
    convert = 10**6
  elif unit == "GB":
    convert = 10**9
  elif unit == "TB":
    convert = 10**12
  elif unit == "KiB":
    convert = 2**10
  elif unit == "MiB":
    convert = 2**20
  elif unit == "GiB":
    convert = 2**30
  elif unit == "TiB":
    convert = 2**40
  else:
    return "Invalid data"
  actualData = task3_2(quantity_of_data)
  if actualData != "Invalid data":
    final = actualData / convert
    return final
  else:
    return "Invalid data"

# main
print(task3_3("512MiB", "GiB"))
print(task3_3("512Mi", "GiB"))
print(task3_3("512MB", "GB"))
````
